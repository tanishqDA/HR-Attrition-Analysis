# **Univariate Analysis**

```python
numerical_columns=df.select_dtypes(include=["int64","float64"]).columns
categorical_columns=df.select_dtypes(include=["object"]).columns
print("Numerical Columns:")
print(numerical_columns)
print("\nCategorical Columns:")
print(categorical_columns)
```

```python
num_cols=[
    "Age",
    "DistanceFromHome",
    "MonthlyIncome",
    "YearsAtCompany",
    "TotalWorkingYears"
]
```

```python
for col in num_cols:
  print(f"\n----- {col} -----")
  print(df[col].describe())
  print("Mean:", round(df[col].mean(),2))
  print("Median:", df[col].median())
  print("Mode:", df[col].mode()[0])
  sb.histplot(df[col],kde=True)
  plt.title(f"Distribution of {col}")
  plt.show()
  sb.boxplot(y=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
cat_col=[
    "Gender",
    "Attrition",
    "Department",
    "JobRole",
    "MaritalStatus",
    "OverTime",
    "Education",
    "WorkLifeBalance"
]
```

```python
for col in cat_col:
  print(f"\n{'='*50}")
  print(f"Analysis of {col}")
  print(f"{'='*50}")
  print("\nFrequency count:")
  print(df[col].value_counts())
  print("\nPercentage:")
  print((df[col].value_counts(normalize=True)*100).round(2))
  sb.countplot(data=df,x=col,order=df[col].value_counts().index)
  plt.title(f"Distribution of {col}")
  plt.xticks(rotation=60)
  plt.show()
```

# **Observation and Buisness Insight from Bivariate Analysis**

1. Attrition vs OverTime

    Observation:

    -Employees who work overtime exhibit a higher level of attrition compared to employees who do not work overtime.

    -The majority of employees who do not work overtime tend to remain with the organization.

    Buisness Insight:

    -Overtime appears to be a major contributor to employee attrition.

2. Attrition vs MonthlyIncome

    Observation:

    -Employees who left the organization generally have lower monthly incomes compared to employees who stayed.

    -The median monthly income of retained employees is noticeably higher than that of employees who left.

    -Several high-income outliers are observed, particularly among employees who remained with the company

    Buisness Insight:

    -Compensation appears to be associated with employee retention, as employees with lower salaries exhibit higher attrition levels.

    -Reviewing salary structures and compensation policies may help improve employee retention, especially among lower-income employees.

3. Attrition vs Age

    Observation:

    -Employees who left the organization tend to be younger compared to employees who stayed.

    -The median age of retained employees is higher than that of employees who left.

    Buisness Insight:

    -Younger employees appear to be more vulnerable to attrition than older employees.

    -Early-career employees may require additional engagement, career development opportunities, and retention initiatives to improve long-term employee retention.

4. Attrition vs JobRole

    Observation:

    -Sales Representatives exhibit the highest attrition rate among all job roles.

    -Research Directors exhibit the lowest attrition rate among all job roles.

    Buisness Insight:

    -Employees in sales-oriented roles appear more likely to leave the organization compared to employees in leadership and research-focused roles.

    -The high attrition rate among Sales Representatives suggests potential challenges related to workload, performance expectations, compensation structure, or career progression

5. Attrition vs Department

    Observation:

    -Although Research & Development records the highest number of employee exits, it also has the largest workforce.

    -When attrition is analyzed as a percentage of department size, the Sales department exhibits the highest attrition rate.

    -Human Resources shows the lowest attrition rate among all departments.

    Buisness Insight:

    -Employees in the Sales department appear more likely to leave the organization compared to employees in other departments.


6. Attrition vs WorkLifeBalance

    Observation:

    -Employees reporting the lowest WorkLifeBalance rating (1) exhibit the highest attrition rate.

    -Attrition tends to decrease as employees report better work-life balance.

    Buisness Insight:

    -Work-life balance appears to be strongly associated with employee retention.

    -Improving workload management, flexibility, and employee well-being initiatives may help reduce attrition

7. Attrition vs MaritalStatus

    Observation:

    -Single employees exhibit the highest attrition rate among all marital status groups.

    -Divorced employees exhibit the lowest attrition rate.

    Buisness Insight:

    -Single employees appear more likely to leave the organization compared to married and divorced employees.

    -Employee retention strategies may need to focus more on younger and single employees who are at a greater risk of attrition.
